## 6. Can you plot the distribution of the durations?

## 7. [Follow-up] Can you make this plot more readable by removing outliers (extreme values)?

```python
# Your code goes here 💪


# Original plot (for comparison)
plt.figure(figsize=(10, 4))
sns.histplot(df['duration_ms'] / 60000, bins=50)
plt.xlabel('Duration (minutes)')
plt.show()

# Without outliers: keep songs shorter than 10 minutes
# (try 600_000 ms; adjust until you remove about 470 songs)
df_no_outliers = df[df['duration_ms'] < 600_000]
print(len(df) - len(df_no_outliers))  # number of songs removed

plt.figure(figsize=(10, 4))
sns.histplot(df_no_outliers['duration_ms'] / 60000, bins=50)
plt.xlabel('Duration (minutes)')
plt.show()
```

We only remove 470 songs (out of almost 170k) and get a way more interesting plot!

```python
# Your code goes here 💪
total_ms = df['duration_ms'].sum()
total_days = total_ms / 1000 / 60 / 60 / 24
print(round(total_days, 1))
```

## 14. [Follow-up] can you make a bar chart out of this result?

## 15. How many songs by key?

## 16. [Follow-up] Can you plot a bar chart using Seaborn? Make sure the keys are sorted in a descending order.

```python
# Your code goes here 💪
key_counts = df['key'].value_counts()   # already sorted descending by count

plt.figure(figsize=(10, 5))
sns.barplot(x=key_counts.index, y=key_counts.values, order=key_counts.index)
plt.title('Number of songs by key')
plt.xlabel('Key')
plt.ylabel('Number of songs')
plt.show()
```

## 17. Can you plot a line graph representing the annual number of songs since 1921?

```python
# Your code goes here 💪
df.groupby('year').size().plot(kind='line', figsize=(12, 5))
plt.title('Number of songs per year')
plt.xlabel('Year')
plt.ylabel('Number of songs')
plt.show()
```
